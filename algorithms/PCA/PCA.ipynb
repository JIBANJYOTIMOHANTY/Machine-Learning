```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X, y= make_blobs(n_samples=500, n_features=5, centers=3, cluster_std=1.5, random_state=42)
```

```python
scaler = StandardScaler()
X_sclaed = scaler.fit_transform(X)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_sclaed)
```

```python
X_pca
```

```python
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
df_pca['label'] = y
```

```python
sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='label', palette='tab10')
```
