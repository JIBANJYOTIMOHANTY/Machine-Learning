```python
leakage_columns = [
    'Churn Score',
    'Churn Category',
    'Churn Reason',
    'Customer Status'
]

df_model = df.drop(columns=leakage_columns)
df_model.shape
```

```python
df_model['TenureGroup'] = pd.cut(
    df_model['Tenure in Months'],
    bins=[0, 12, 24, 48, 72],
    labels=['0-1 Year', '1-2 Years', '2-4 Years', '4-6 Years'],
    include_lowest=True
)
df_model[['Tenure in Months', 'TenureGroup']].head()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df_model, x='TenureGroup', hue='Churn Label')
plt.title('Churn By Tenure Group')
plt.show()
```

```python
churn_rate = pd.crosstab(
    df_model['TenureGroup'],
    df_model['Churn Label'],
    normalize='index'
) * 100

churn_rate
```

```python
df_model['NewCustomer']=(
    df_model['Tenure in Months']<=12
    ).astype(int)
df_model['NewCustomer'].head()
```
