```python
sns.countplot(data=df, x = 'Churn Label')
plt.title("Customer Churn Distribution")
plt.show()
```

```python
sns.countplot(data=df, x = 'Contract', hue='Churn Label')
plt.xticks(rotation=45)
plt.title('Churn By Contract Type')
plt.show()
```

```python
sns.boxplot(data=df, x = 'Churn Label', y='Tenure in Months')
plt.title('Tenure Vs Churn')
plt.show()
```

```python
sns.boxplot(data=df, x='Churn Label', y='Monthly Charge')
plt.title('Monthly Charge vs Churn')
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(
    df.select_dtypes(include=['number']).corr(),
    cmap='coolwarm',
    annot=True
)
plt.title("Numerical Feature Correlation")
plt.show()
```

```python
df.groupby('Churn Label')['Tenure in Months'].median()
```

```python
sns.countplot(
    data=df,
    x='Internet Service',
    hue='Churn Label'
)

plt.title('Churn by Internet Service')
plt.xticks(rotation=45)
plt.show()
```
