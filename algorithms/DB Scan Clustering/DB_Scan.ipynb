```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
X, y_true = make_moons(n_samples=500, noise=0.05, random_state=42)
```

```python
df = pd.DataFrame(X, columns=['Feature 1', 'Feature 2'])
df.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['kmenas_cluster'] = kmeans_labels
```

```python
sns.scatterplot(x=df['Feature 1'], y=df['Feature 2'], hue=df['kmenas_cluster'], palette='tab10')
```

```python
db_scan = DBSCAN(eps=0.3, min_samples=5)
db_scan_labels = db_scan.fit_predict(X_scaled)
```

```python
df['db_scan_cluster'] = db_scan_labels
```

```python
sns.scatterplot(x=df['Feature 1'], y=df['Feature 2'], hue=df['db_scan_cluster'], palette='tab10')
```
