```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5, 11)
x
y = x ** 2
y
```

```python
plt.title("Learning Matplotlib")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.plot(x, y)
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
```

```python
plt.subplot(2,2,2)
plt.plot(x,y**3)
plt.subplot(2,2,3)
plt.plot(x**2,x**2)
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.4,0.3,0.3])
axis2.plot(x,y)
plt.savefig('basicFig.png')
```
