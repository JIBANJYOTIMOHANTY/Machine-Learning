```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('IPL.csv')
df
```

```python
df.info()
```

```python
df.shape
```

```python
print('Total number of rows are : ', df.shape[0])
print('Total number of columns are : ', df.shape[1])
print('Total number of null values in each row')
df.isnull().sum()
```

```python
match_wins = df['match_winner'].value_counts()
match_wins
```

```python
sns.barplot(y = match_wins.index, x = match_wins.values)
plt.title("Most matches win by the team")
plt.xlabel("Total number of match wins")
plt.ylabel("Total Number Of Teams")
```

```python
sns.countplot(x = df['toss_decision'])
df['toss_decision'].value_counts()
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100) / df.shape[0]
print("Total number of match winners by toss winners",count)
print("Percentage",percentage)
```

```python
sns.countplot(x = df['won_by'])
```

```python
player_of_the_matches = df['player_of_the_match'].value_counts().head(10)
player_of_the_matches
player_of_the_matches
```

```python
sns.barplot(x = player_of_the_matches.values, y = player_of_the_matches.index)
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high
high.plot(kind='barh')
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(5)
top_bowlers
```

```python
top_bowlers.plot(kind='barh')
```
