```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "transaction_amount" : [500, 700, np.nan, 600, 900, np.nan, 800, 1000],
    "transaction_time" : [10, 11, 12, np.nan, 14, 15, 16, 17]
})

print("Before Imputation : ")
print(df)

amount_median = df['transaction_amount'].median()
time_median = df['transaction_time'].median()

df['transaction_amount'] = df['transaction_amount'].fillna(amount_median)
df['transaction_time'] = df['transaction_time'].fillna(time_median)

print("\nAfter Imputation")
print(df)

print("\n",df.isnull().sum())
```
