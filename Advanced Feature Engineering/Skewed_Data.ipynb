```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

amounts = np.array([
    100, 150, 200, 250, 300,
    350, 400, 450, 500, 600,
    700, 800, 900, 1000, 1500,
    2000, 5000, 10000, 250000, 100000
])

log_amounts = np.log1p(amounts)

comparison = pd.DataFrame({
    "Original_Amount" : amounts,
    "Log_Amount" : log_amounts
})

print(comparison.round(3))
print("\nOriginal Skewness : ", pd.Series(amounts).skew().round(3))
print("Log-transformed skewness : ", pd.Series(log_amounts).skew().round(3))
```

```python
plt.figure(figsize=(8,4))
plt.hist(amounts, bins=10, edgecolor="black")
plt.title("Original Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.hist(log_amounts, bins=10, edgecolor="black")
plt.title("Log-Transformed Trasaction Amounts")
plt.xlabel("log(1 + Transaction Amounts)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
