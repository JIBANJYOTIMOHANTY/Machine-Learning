```python
import pandas as pd

df = pd.DataFrame({
    "customer_id" : ["A", "A", "A", "B", "B"],
    "transaction_amount" : [500, 600, 5000, 300, 400],
    "transaction_hour" : [10, 11, 2, 14, 15]
})

df["customer_avg_amount"] = df.groupby('customer_id')['transaction_amount'].transform("mean")
df['amount_ratio'] = df['transaction_amount'] / df['customer_avg_amount']
df['is_early_morning'] = (df['transaction_amount'] < 6).astype(int)

print(df)
```
