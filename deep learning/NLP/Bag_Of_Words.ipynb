```python
from sklearn.feature_extraction.text import CountVectorizer

sentences = [
    "I love Python",
    "I love Deep Learning",
    "Python is intresting"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(sentences)

print("Vocabulary")
print(vectorizer.get_feature_names_out())

print("BOW Matrix")
print(X.toarray())
```
