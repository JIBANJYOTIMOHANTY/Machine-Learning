```python
from sklearn.feature_extraction.text import CountVectorizer

sentences = [
    "I love Python",
    "I love Deep Learning",
    "Python is intresting"
]

unigram = CountVectorizer(ngram_range=(1, 1))
X_uni = unigram.fit_transform(sentences)

print("Unigram Vocabulary : ")
print(unigram.get_feature_names_out())

print("Unigram Matrix")
print(X_uni.toarray())
```

```python
bigram = CountVectorizer(ngram_range=(2, 2))
X_bi = bigram.fit_transform(sentences)

print("\nBigram Vocabulary : ")
print(bigram.get_feature_names_out())

print("\nBigram Matrix")
print(X_bi.toarray())
```

```python
combined = CountVectorizer(ngram_range=(1, 2))
X_combined = combined.fit_transform(sentences)

print("\nCombined Vocabulary : ")
print(combined.get_feature_names_out())

print("\nCombined Matrix")
print(X_combined.toarray())
```
