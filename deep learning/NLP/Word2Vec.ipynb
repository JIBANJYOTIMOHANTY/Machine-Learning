```python
import pandas as pd
import numpy as np
import re
import tensorflow as tf
```

```python
sentences = [
    "I love machine learning",
    "I love deep learning",
    "machine learning is interesting",
    "deep learning is powerful",
    "python is used for machine learning",
    "python is used for deep learning",
    "neural networks are part of deep learning",
    "artificial intelligence uses machine learning",
    "data science uses python",
    "machine learning uses data",
    "deep learning uses neural networks",
    "natural language processing uses deep learning",
    "word embeddings represent words as vectors",
    "word2vec learns word embeddings",
    "machine learning algorithms learn from data",
    "neural networks learn patterns from data",
    "python is a popular programming language",
    "artificial intelligence is an exciting field"
]
```

```python
tokenized_sentences = []

for sentence in sentences:
    sentence = sentence.lower()
    sentence = re.sub(r"[^a-z0-9\s]", "", sentence)
    words = sentence.split()
    tokenized_sentences.append(words)

print("Tokenized Sentences")
print(tokenized_sentences)
```

```python
vocabulary = sorted(
    set(
        word
        for sentence in tokenized_sentences
        for word in sentence
    )
)

word_to_index = {
    word : index
    for index, word in enumerate(vocabulary)
}

index_to_word = {
    index : word
    for word, index in word_to_index.items()
}

vocab_size = len(vocabulary)

print("\nVocabulary size:", vocab_size)
print("\nVocabulary:")
print(vocabulary)
```

```python
window_size = 2
X_train = []
y_train = []

for sentence in tokenized_sentences:
    for i, center_word in enumerate(sentence):
        center_index = word_to_index[center_word]
        start = max(0, i - window_size)
        end = min(len(sentence), i + window_size + 1)

        for j in range(start, end):
            if i != j:
                context_word = sentence[j]
                context_index = word_to_index[context_word]
                X_train.append(center_index)
                y_train.append(context_index)
```

```python
X_train = np.array(X_train, dtype=np.int32)
y_train = np.array(y_train, dtype=np.int32)

print("\nTraining input shape:", X_train.shape)
print("Training output shape:", y_train.shape)
```

```python
print("\nFirst 10 training pairs:")

for i in range(10):
    print(index_to_word[X_train[i]], "->", index_to_word[y_train[i]])
```

```python
embedding_dim = 50

model = tf.keras.Sequential([
    tf.keras.Input(shape=(), dtype=tf.int32),

    tf.keras.layers.Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim
    ),

    tf.keras.layers.Dense(
        vocab_size,
        activation='softmax'
    )
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    X_train,
    y_train,
    epochs=200,
    batch_size=16,
    verbose=1
)
```

```python
embedding_layer = model.layers[0]
word_embeddings = embedding_layer.get_weights()[0]
print(word_embeddings.shape)

word = "learning"
word_index = word_to_index[word]
word_vector = word_embeddings[word_index]

print("\nWord:", word)

print("\nWord Vector:")
print(word_vector)

print("\nVector Shape:")
print(word_vector.shape)
```

```python
def find_similar_words(word, top_n=5):

    if word not in word_to_index:
        print("Word not found in vocabulary!")
        return

    word_index = word_to_index[word]

    target_vector = word_embeddings[word_index]

    similarities = np.dot(
        word_embeddings,
        target_vector
    ) / (
        np.linalg.norm(word_embeddings, axis=1)
        * np.linalg.norm(target_vector)
        + 1e-10
    )

    similar_indices = np.argsort(similarities)[::-1]

    print("\nWords similar to:", word)

    count = 0

    for index in similar_indices:

        if index == word_index:
            continue

        print(
            index_to_word[index],
            ":",
            round(float(similarities[index]), 4)
        )

        count += 1

        if count == top_n:
            break


find_similar_words("learning", top_n=5)

np.save(
    "word2vec_embeddings.npy",
    word_embeddings
)

print("\nWord2Vec embeddings saved successfully!")
```
