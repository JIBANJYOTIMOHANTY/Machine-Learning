```python
import numpy as np

Q = np.array([1, 0])

K = np.array([
    [1, 0],
    [0, 1]
])

V = np.array([
    [10, 20],
    [30, 40]
])

scores = Q @ K.T
scores
```

```python
exp_scores = np.exp(scores)
attention_weights = exp_scores / np.sum(exp_scores)

output = attention_weights @ V
print("Scores : ", scores)
print("Attention Weights : ", attention_weights)
print("Attention Output : ", output)
```
