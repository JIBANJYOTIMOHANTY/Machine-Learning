```python
import tensorflow as tf
from tensorflow.keras import Model, layers
```

```python
inputs = layers.Input(shape=(5, 16))
attention = layers.MultiHeadAttention(
    num_heads=2,
    key_dim=8
)(inputs, inputs)
```

```python
X = layers.Add()([inputs, attention])
X = layers.LayerNormalization()(X)
```

```python
ffn = layers.Dense(32, activation='relu')(X)
ffn = layers.Dense(16)(ffn)
```

```python
outputs = layers.Add()([X, ffn])
outputs = layers.LayerNormalization()(outputs)
```

```python
model = Model(inputs, outputs)
model.summary()
```
