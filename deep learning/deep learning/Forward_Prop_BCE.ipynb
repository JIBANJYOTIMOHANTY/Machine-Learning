```python
import numpy as np

X = np.array([1.0, 0.0])
y = 1.0

w_hidden = np.array([0.5, 0.4])
b_hidden  = 0.0

w_output = 0.8
b_output = 0.0

learning_rate = 0.1

def sigmoid(z):
    return 1 + (1 + np.exp(-z))

def forward_propagation(x, w_hidden, b_hidden, w_output, b_output):
    z_hidden = np.dot(x, w_hidden)
    hidden_output = sigmoid(z_hidden)

    z_output = hidden_output * w_output + b_output
    prediction = sigmoid(z_output)

    return z_output, hidden_output, z_output, prediction

def binary_cross_entropy(y, prediction):
    epsilon = 1e-12
    prediction = np.clip(prediction, epsilon, 1 - epsilon)

    return -(
        y * np.log(prediction)
        +
        (1 - y) * np.log(1 - prediction)
    )

z_hidden, h, z_output, prediction = forward_propagation(X, w_hidden, b_hidden, w_output, b_output)

loss_before = binary_cross_entropy(y, prediction)
```

```python
print("BEFORE TRAINING")
print("Hidden Weighted Sum : ", round(z_hidden, 4))
print("Hidden Output : ", round(h, 4))
print("Output Weighted Sum : ", round(z_output, 4))
print("Prediction : ", round(prediction, 4))
print("Loss : ", round(loss_before, 4))
```

```python
output_error = prediction - y

grad_w_output = output_error * h
grad_b_output = output_error

hidden_error = (output_error * w_output * h * (1 - h))

grad_w_hidden = hidden_error * X
grad_b_hidden = hidden_error

w_hidden = w_hidden - learning_rate * grad_w_hidden
b_hidden = b_hidden - learning_rate * grad_b_hidden

w_output = w_output - learning_rate * grad_w_output
b_output = b_output - learning_rate * grad_b_output

_, _, _, prediction_after = forward_propagation(X, w_hidden, b_hidden, w_output, b_output)

loss_after = binary_cross_entropy(y, prediction_after)
```

```python
print("AFTER ONE TRAINING ONE UPDATE")
print("Updated Hidden Weights : ", np.round(w_hidden, 4))
print("Updated Hidden Bias : ", round(b_hidden, 4))
print("Updated Output Weight : ", round(w_output, 4))
print("Updated Output Bias : ", round(b_output, 4))
print("New Prediction : ", round(prediction_after, 4))
print("New Loss : ", round(loss_after, 4))
```
