```python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def tanh(z):
    return np.tanh(z)

def relu(z):
    return np.maximum(0, z)

def softmax(logits):
    shifted_logits = logits - np.max(logits)
    exp_values = np.exp(shifted_logits)
    return exp_values / np.sum(exp_values)
```

```python
z_sigmoid = 2.0
sigmoid_output = sigmoid(z_sigmoid)
print("Sigmoid : ", sigmoid_output)

z_tanh = 1.0
tanh_output = tanh(z_tanh)
print("Tanh : ", tanh_output)

relu_inputs = np.array([-3.0, 0.0, 4.0])
relu_outputs = relu(relu_inputs)
print("ReLU", relu_outputs)
```

```python
logits = np.array([2.0, 1.0, 0.1])
probabilities = softmax(logits)
classes = ["Cat", "Dog", "Bird"]

print("SoftMax")
for class_name, probability in zip(classes, probabilities):
    print(f"{class_name} : {probability:.4f}")

print("Probability Sum : ", round(float(np.sum(probabilities)), 4))
predicted_class = classes[np.argmax(probabilities)]
print("Predicted Class : ", predicted_class)
```

```python
test_inputs = np.array([-2.0, 0.0, 2.0])
print("ACTIVATION COMPARISION")
print("Inputs : ", test_inputs)
print("Sigmoid : ", np.round(sigmoid(test_inputs), 4))
print("Tanh : ", np.round(tanh(test_inputs), 4))
print("ReLU : ", relu(test_inputs))
```
