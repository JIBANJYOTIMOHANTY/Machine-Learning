```python
import numpy as np

X = np.array([4.0, 2.0])
weights = np.array([0.6, 0.3])

bias = -1.0

z = np.dot(X, weights) + bias

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

output = sigmoid(z)

prediction = 1 if output >= 0.5 else 0

print("Weighted Sum : ", z)
print("Sigmoid Output : ", output)
print("Binary Prediction : ", prediction)
```
